```python
# 1-exercise
def is_prime(a:int):
    if a>1 and type(a)==int:
        for i in range(2,int(a**0.5)+1):
            if a%i==0:
                return False
        return True  
    else:
        print('You wrote incorrect datatype ?')
is_prime(8.2)
```

```python
#3-exercise
def roman_to_number(roman):
    roman_map = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    result = 0
    next_value = 0
    for char in reversed(roman):
        value = roman_map[char]
        if value < next_value:
            result -= value
        else:
            result += value
        next_value = value
    return result

roman_to_number('XXVII')

```
