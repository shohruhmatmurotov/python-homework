```python
def modify_string(txt):
    vowels = "aeiou"
    result = []
    count = 0

    for i, char in enumerate(txt):
        result.append(char)
        if char not in vowels:
            count += 1
        if count % 3 == 0 and i != len(txt) - 1:
            result.append("_")

    return "".join(result)
```

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


is_prime(25
         )
```
