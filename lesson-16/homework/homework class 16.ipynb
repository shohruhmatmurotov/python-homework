```python
#56
import scipy.stats as st

def gaussian_2d(shape=(10,10), sigma=1):
    m, n = [(dim-1)/2 for dim in shape]
    y, x = np.ogrid[-m:m+1, -n:n+1]
    return np.exp(-(x**2 + y**2) / (2 * sigma**2))

gaussian = gaussian_2d()
print(gaussian)
```
