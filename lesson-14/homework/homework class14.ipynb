```python
#5. Create a list of destinations that you offer flights with the distance
# Ask for the user the day they are flying
# If it is a weekend offer them 15% discount
# If the distance is greater than 1500, charge them 20% more of the base price
# You will define your own base price
# and ask what class do they want to fly, if it is business, price + 40%
from datetime import date,datetime
def travelling():
    
    destinations = {
        "New York": 1600,
        "Los Angeles": 3000,
        "Chicago": 1200,
        "Miami": 1400,
        "London": 7000,
        "Dubai": 5000,
        "Urganch":1000,
        "Munich":3000
    }
    base_price=700

    print('Available tours:{destinations}')
    for d,i in destinations.items():
        print( {d}-{i})
    location=input('write the location ...from list of tours')
    date=input('write the date like : dd.mm.yyyy')
    choice=input('do you want to fly in bussiness class ,if you want write yes').lower()
    weekday=datetime.strptime(date,'%d.%m.%Y').weekday
    if location in destinations:
       
        distance = destinations[location]
        price = base_price

    
        if distance > 1500:
            price *= 1.2  

    
        if weekday==5 or weekday==6: 
            price *= 0.85  

    
        if choice == "yes":
            price *= 1.4  
        print(f'your tour for {location} on {date}:💲{price}')
    else:
        print('incorrect location!')

        
    travelling()
```

```python
#Task3
import json
import os
books_file = 'books.json'

def load_books():
    if os.path.exists(books_file):
        with open(books_file, 'r') as file:
            return json.load(file)
    return []

def save_books(books):
    with open(books_file, 'w') as file:
        json.dump(books, file, indent = 4)

def add_book():
    books = load_books()
    title = input('Enter book title:')
    author = input('Enter the author')
    year = input('Enter the year of book release')
    if books['title']!=title:
        book = {'title':title, 'author':author, 'year':year}
        books.append(book)
        save_books(books)
        print('Book added successfully')
    else:
        print('This books has added yet in our library !')
def update_book():
    """Update book details by searching for the title."""
    books = load_books()
    print(books)
    
    title = input("\nEnter the title of the book to update: ").strip()

    for book in books:
        if book["title"].lower() == title.lower():
            book["author"] = input(f"Enter new author (current: {book['author']}): ").strip() or book["author"]
            book["year"] = input(f"Enter new year (current: {book['year']}): ").strip() or book["year"]
            save_books(books)
            print(" Book updated successfully!")
            return

    print("Book not found!")

def delete_book():
    
    books = load_books()
    
    
    title = input("\nEnter the title of the book to delete: ").strip()

    new_books = [book for book in books if book["title"].lower() != title.lower()]
    
    if len(new_books) == len(books):
        print(" Book not found!")
        return

    save_books(new_books)
    print("Book deleted successfully!")
def main():
    'This instraction of our program:'
    while True:
        print(" Book Manager")
        print("1. Display Books")
        print("2. Add Book")
        print("3. Update Book")
        print("4. Delete Book")
        print("5. Exit")
        choice=int(input('write the number of function you want to use!'))
        if choice==1:
            load_books()
        if choice==2:
            add_book()
        if choice==3:
            update_book()
        if choice==4:
            delete_book()
        if choice==5:
            break

    
```
