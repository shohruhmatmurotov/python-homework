```python
import pandas as pd 

sales_data=pd.read_csv(r"C:\Users\Asus\Downloads\sales_data.csv")

customer_orders=pd.read_csv(r"C:\Users\Asus\Downloads\customer_orders.csv")
```

```python
#Homework-1
sales_data.groupby('Category').agg({'Quantity':['sum','mean'],'Price':'mean'})

sales_data.groupby('Category').agg({'Quantity':'sum'}).sort_values(ascending=False,by='Quantity').head(1)

sales_data['totalsales']=sales_data['Quantity']*sales_data['Price']
sales_data.sort_values(by='totalsales',ascending=False).max().Date
```

```python
#Homework-2
cust=customer_orders.groupby('CustomerID').agg({'OrderID':'count'})
cust[cust['OrderID']>20]

data=customer_orders.groupby(['CustomerID',]).agg({'Price':'mean'})
data[data['Price']>120]

total=customer_orders.groupby('Product').agg({'Price':'sum','Quantity':'sum'})
total[total['Quantity']<5]
```
